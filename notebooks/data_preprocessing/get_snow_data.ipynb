## Chena-area snow

Download SWE and snow depth for the Chena snow courses, SNOTEL stations, and NOHRSC airborne flight lines. Values are written in millimetres. Edit the settings cell, then run the download. The same download can be run from `scripts/get_snow_data.py`.

```python
from pathlib import Path
import sys

import pandas as pd

repo = Path.cwd().resolve()
while not (repo / "pyproject.toml").exists() and repo != repo.parent:
    repo = repo.parent
sys.path.insert(0, str(repo / "scripts"))

from get_snow_data import download_snow

output_dir = repo / "outputs"
output_dir.mkdir(parents=True, exist_ok=True)
```

## Settings

`start` and `end` are inclusive dates (`YYYY-MM-DD`). Leave `end` as `None` to use today.

`sources` selects which networks to download:

- `snotel` — daily SNOTEL SWE and snow depth
- `nrcs` — manual snow-course surveys
- `nohrsc` — airborne SWE (skipped unless `WTEQ` is requested)

`elements` is `WTEQ` (snow water equivalent) and/or `SNWD` (snow depth). `refresh` ignores cached NRCS and NOHRSC responses. A long window requests one year at a time and can take a while.

```python
start = "1980-01-01"
end = None

out = output_dir / "chena_snow"
sources = ["snotel", "nrcs", "nohrsc"]
#elements = ["WTEQ", "SNWD"]
elements = ["SNWD"]
refresh = False
```

```python
result = download_snow(
    start=start,
    end=end,
    out=out,
    sources=sources,
    elements=elements,
    refresh=refresh,
)

values = result.data["value_mm"].notna()
print(
    f"{len(result.data)} rows ({int(values.sum())} with values) "
    f"-> {result.out / 'snow_all_mm.csv'}"
)
if not result.ok:
    print("Download incomplete; see the status table.")
result.data.head()
```
