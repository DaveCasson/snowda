## Historic daily discharge

Download the complete daily discharge record for one gauge and write a CSV. Discharge is the daily mean in m3/s. The same download can be run from `scripts/get_canadian_discharge.py`.

```python
from pathlib import Path
import sys

import pandas as pd

repo = Path.cwd().resolve()
while not (repo / "pyproject.toml").exists() and repo != repo.parent:
    repo = repo.parent
sys.path.insert(0, str(repo / "scripts"))

from get_wsc_discharge import fetch_wsc_daily_means, write_wsc_csv
from get_usgs_discharge import fetch_usgs_daily_means, write_usgs_csv

output_dir = repo / "outputs"
output_dir.mkdir(parents=True, exist_ok=True)
```

## Canada: Ross River at Ross River (09BA001)

Water Survey of Canada daily means from the MSC GeoMet `hydrometric-daily-mean` collection. `discharge_symbol` is the HYDAT flag (for example ice conditions or estimated).

```python
station = "09BA001"
name = "ross"
path = output_dir / f"{name}_daily_discharge.csv"
rows = fetch_wsc_daily_means(station)
write_wsc_csv(rows, path)

data = pd.read_csv(path, parse_dates=["date"])
print(
    f"{data['station_name'].iloc[0]} ({station}): "
    f"{data['date'].min().date()} to {data['date'].max().date()}, "
    f"{int(data['discharge_cms'].notna().sum())} days with discharge -> {path}"
)
data.head()
```

## Chena River near Two Rivers

USGS site `15493000` is a separate example: `notebooks/data_preprocessing/get_chena_discharge.ipynb`, or `python3 scripts/get_usgs_discharge.py 15493000`.
