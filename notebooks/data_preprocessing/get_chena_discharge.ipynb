## Chena River near Two Rivers

Download the complete USGS daily discharge record for Chena River near Two Rivers, Alaska (site `15493000`) and write a CSV. Discharge is the daily mean converted from ft3/s to m3/s. `discharge_symbol` is the NWIS qualifier (`A` approved, `P` provisional, `e` estimated). The same download can be run from `scripts/get_usgs_discharge.py`.

```python
from pathlib import Path
import sys

import pandas as pd

repo = Path.cwd().resolve()
while not (repo / "pyproject.toml").exists() and repo != repo.parent:
    repo = repo.parent
sys.path.insert(0, str(repo / "scripts"))

from get_usgs_discharge import fetch_daily_means, write_csv

output_dir = repo / "outputs"
output_dir.mkdir(parents=True, exist_ok=True)
```

```python
site = "15493000"
chena_path = output_dir / f"{site}_daily_discharge.csv"
chena_rows = fetch_daily_means(site)
write_csv(chena_rows, chena_path)

chena = pd.read_csv(chena_path, parse_dates=["date"])
print(
    f"{chena['station_name'].iloc[0]} ({site}): "
    f"{chena['date'].min().date()} to {chena['date'].max().date()}, "
    f"{int(chena['discharge_cms'].notna().sum())} days with discharge -> {chena_path}"
)
chena.head()
```
