# Run modes

Pysummada configs now use `run_mode` to select the workflow behavior:
- `deterministic`: no DA, no ESP
- `deterministic_esp`: no DA, ESP using deterministic state (`esp.deterministic` or `esp.deterministic_state`)
- `da`: DA only, no ESP
- `da_esp`: DA followed by ESP using DA states
